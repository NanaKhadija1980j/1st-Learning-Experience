### **Conditional Statements in Python (`Control Flow`)**

##### **What is a Conditional Statement?**
**Conditional statements** allow your program to make decisions and run specific blocks of code based on whether a condition is **True** or **False**.

A conditional statement is a way to tell your program:

> “If a condition is true, do one thing; if not, do something else.”

## Learning Objectives

By the end of this lesson, you will be able to:

- Understand and use **comparison operators**.
- Write simple conditionals using the `if` statement
- Use **branching** with `if`, `elif`, and `else` to handle multiple conditions
- **Nest conditionals** inside each other to handle more complex logic
- Apply **logical operators** (`and`, `or`, `not`) to combine multiple conditions
- Apply conditionals to real-world problems and simple data science tasks

## Comparison Operators

## Simple Conditionals: `if` Statement

Creating a conditional statement allows your code to branch into two or more different paths.

The `if` statement is how Python decides **what to do** if a condition is true.

**Syntax**:

```python
if condition:
    # do something
```

````python
# Demonstrating conditionals by monitoring a the water tank

Water_tank_weight = 13

if Water_tank_weight == 20:
    print("The water tank is full! We are turning off the pump")
````

## Branching with `if`, and `else`

This is used when you have only 2 action. **If** `X` happens, do `A`  **else** do `B`.

**Syntax**:

```python
if condition:
    # do this
else:
    # do something else
```

````python
# Demonstrating conditionals by monitoring a the water tank

# Define the variable
Water_tank_weight = 15

# Compare the values
if Water_tank_weight == 20:
    print("The water tank is full! We are turning off the pump")
else:
    print("The tank is not yet full")
````

## Branching with `if`, `elif`, and `else`

You can check multiple conditions using `elif` (short for *else if*) and `else`.

**Syntax**:

```python
if condition1:
    # do this
elif condition2:
    # do that
else:
    # do something else
```

````python
# Movie recommender based on age category

# less than 12 --- Recommend Tom and Jerry
# less than 18 --- Recommend Avengers
# 18 and above --- News
````

````python
Age = 15

if Age < 12:
    print("Hello, you can watch Tom and Jerry")
elif Age < 18:
    print("Hello, you can watch Avengers")
else:
    print("Hello, you can enjoy the News")
````

````python
Age = int(input("Please enter your age: "))  # Casting int on the input

if Age < 12:
    print("Hello, you can watch Tom and Jerry")
elif Age >= 12 and Age < 18:
    print("Hello, you can watch Avengers")
else:
    print("Hello, you can enjoy the News")
````

## Problem Statement: Event Entry & Recommendation System with Signup

### Objective:

Create a Python program that **verifies event attendees** using a simple database and gives them **personalized recommendations** based on their age. If a user is not in the database, offer them the option to **sign up immediately**.

### Background:

You’re building a basic event system for a concert. The system stores user data (name, age, ticket number) in a dictionary, and it checks whether a visitor is registered. If the person isn’t found, they should be given the option to **sign up** on the spot.

After verifying or registering a user, the system will suggest the best part of the event for them based on their age.

### Step-by-step Requirements:

1. **Create an initial database** (dictionary) with at least 5 users.
   Each user should have:

   * A name (as the key)
   * A value that stores both `age` and `ticket number`

   Example:

   ```python
   database = {
       "Ifeoma": {"age": 25, "ticket": "TCK001"},
       "Amanda": {"age": 16, "ticket": "TCK002"},
       "Afolabi": {"age": 34, "ticket": "TCK003"},
       "Wale": {"age": 12, "ticket": "TCK004"},
       "Chioma": {"age": 60, "ticket": "TCK005"}
   }
   ```

2. **Collect input from the user**:

   * Name
   * Age
   * Ticket number

3. **Check if the name exists in the database**:

   * If yes: Check if the ticket number matches.

     * If both match → Proceed to age-based recommendation.
     * If not → Show error: *"Ticket number does not match."*
   * If name is **not in the database**, offer them the chance to **sign up**.

     * If they say **yes**, collect their details and add them to the dictionary.

4. **Use conditionals** (`if`, `elif`, `else`, possibly nested) to:

   * Handle the name/ticket verification
   * Handle sign-up
   * Make a recommendation based on age:

   | Age Range | Message                                             |
   | --------- | --------------------------------------------------- |
   | 0–12      | "You're in the Kids Zone!"                          |
   | 13–17     | "You might enjoy the Teen Lounge."                  |
   | 18–35     | "Check out our Main Concert Area!"                  |
   | 36–55     | "VIP Relaxation Lounge available for you."          |
   | 56+       | "We have special seating and assistance if needed." |

````python
# Crate a db
# Collect user input
# Validate age
# Validate name from db
# Validate ticket from db
# Make age based recommendation
````

````python
# Initial user database
database = {
    "Ifeoma": {"age": 25, "ticket": "TCK001"},
    "Wale": {"age": 16, "ticket": "TCK002"},
    "Paul": {"age": 34, "ticket": "TCK003"},
    "Odinaka": {"age": 12, "ticket": "TCK004"},
    "Amanda": {"age": 60, "ticket": "TCK005"}
}
database
````

````python
# Collect user input
name = input("Enter your name: ").strip()
age_input = input("Enter your age: ").strip()
ticket = input("Enter your ticket number: ").strip()
````

````python
# Validate input
if not age_input.isdigit():
    print("Invalid age. Please enter a number.")
else:
    age = int(age_input)

    if name in database:
        if ticket == database[name]["ticket"]:
            print(f"Welcome back, {name}!")

            if age <= 12:
                print("You're in the Kids Zone!")
            elif age <= 17:
                print("You might enjoy the Teen Lounge.")
            elif age <= 35:
                print("Check out our Main Concert Area!")
            elif age <= 55:
                print("VIP Relaxation Lounge available for you.")
            else:
                print("We have special seating and assistance if needed.")
        else:
            print("Ticket number does not match.")
    
    else:
        print(f"The name '{name}' not found in the database.")
        signup = input("Would you like to sign up? (yes/no): ").strip().lower()
        
        if signup == "yes":
            new_age_input = input("Enter your age to sign up: ").strip()
            new_ticket = input("Enter your new ticket number: ").strip()

            if new_age_input.isdigit():
                new_age = int(new_age_input)
                database[name] = {"age": new_age, "ticket": new_ticket}
                print(f"Signup successful! Welcome, {name}!")

                if new_age <= 12:
                    print("You're in the Kids Zone!")
                elif new_age <= 17:
                    print("You might enjoy the Teen Lounge.")
                elif new_age <= 35:
                    print("Check out our Main Concert Area!")
                elif new_age <= 55:
                    print("VIP Relaxation Lounge available for you.")
                else:
                    print("We have special seating and assistance if needed.")
            else:
                print("Invalid age. Signup failed.")
        else:
            print("No access. Please register to proceed.")
````
