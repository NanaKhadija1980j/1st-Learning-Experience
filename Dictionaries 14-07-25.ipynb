### **Introduction to Nested Dictionaries in Python**  

**Nested dictionaries** are dictionaries that contain other dictionaries as values, allowing you to represent hierarchical or structured data in a clean and organized way. They are especially useful for:  

- **Storing multi-level data** (e.g., JSON-like configurations, user profiles, or API responses).  
- **Modeling real-world relationships** (e.g., employees in departments, students in courses).  
- **Simplifying complex data access** with keys instead of lists of lists.  

#### **Key Operations**:  
- **Accessing nested values**: `Parent_dictionary["Child_dictionary"]["Choice_key"]` 
- **Modifying nested data**: `Parent_dictionary["Child_dictionary"]["Choice_key"] = [New_Value]`
- **Modifying nested data by adding new values**: `Parent_dictionary["Child_dictionary"]["New_key"] = [New_Value]` 
- **Iterating through levels**:  Using loops (Advanced).

```python
Position = [1,2,3,4]
Name = ["A","B","C","D"]
```

```python
Student_Dictionary = dict(zip(Name, Position))
```

```python
Student_Dictionary
```

```python
Student_Dictionary
```

```python
type(Student_Dictionary)
```

```python
Grades = [
    ["A", "B"], 
    ["B","C"], 
    ["A","D"]]
Students = ["James", "Peter", "Paul"]
```

```python
Students_Grade_Dict = dict(zip(Grades, Students))
```

```python
Students_Grade_Dict = dict(zip(Students, Grades))
Students_Grade_Dict
```

```python
Grades = [
    {"Chemistry": "A", "Physics" : "B"}, 
    {"Chemistry": "B", "Physics" : "C"}, 
    {"Chemistry": "B", "Physics" : "D"} 
]
Students = ["James", "Peter", "Paul"]
```

```python
Students_complete_dict = dict(zip(Students, Grades))
Students_complete_dict
```

```python
# Using Parent_dictionary["Child_dictionary"]

Students_complete_dict["James"]
```

```python
# Using Parent_dictionary["Child_dictionary"]["Choice_key"]

Students_complete_dict["Peter"]["Chemistry"]
```

#### Modifying Nested Data

```python
# Using Parent_dictionary["Child_dictionary"]["Choice_key"] = [New_Value]

Students_complete_dict["Peter"]["Chemistry"] = ["A"]
```

```python
Students_complete_dict["Peter"]["Chemistry"]
```

```python
Students_complete_dict["Paul"]["Physics"] = ["B"]
```

```python
Students_complete_dict["Paul"]["Physics"]
```

#### **Using the append method**

```python
Students_complete_dict = {
    "Samuel": {
        "Grades": ["B", "C"],
        },
    "Philip": {
        "Grades": ["A", "F"],
        }}
```

```python
Students_complete_dict["Samuel"]["Grades"].append("A")
```

```python
Students_complete_dict["Philip"]["Grades"].append("B")
```

```python
Students_complete_dict["Samuel"]
```

```python
# Storing multi level data

Students_complete_dict = {
    "Samuel": {"Grades": {"Chemistry": "A", "Physics": "B"},},
    "Philip": {"Grades": {"Chemistry": "B"},}}
```

```python
Students_complete_dict["Philip"]["Grades"]["Physics"] = "B"
```

```python
Students_complete_dict["Philip"]["Grades"]
```

```python
Students_complete_dict["Samuel"]["Grades"]["Biology"] = "C"
Students_complete_dict["Philip"]["Grades"]["Biology"] = "A"
Students_complete_dict
```

### **More on accessing dictionaries:** Let's use the `in` and `not in` key words to check if a `key` is in a paricular dictionary.

```python
DS_Students = {
    "James":{"Age": 20, "Salary":33000, "Height": "6ft"},
    "Samuel":{"Age": 22, "Salary":43000, "Height": "5ft"},
    "John":{"Age": 25, "Salary":53000, "Height": "4ft"},
}
```

```python
# Check if we hacve samuel's data stored

"Samuel" in DS_Students
```

```python
# Check if we hacve Anthony's data stored

"Anthony" in DS_Students
```

```python
# Check if we hacve Anthony's data stored

"Anthony" not in DS_Students
```
