### **2. String Methods**

Python includes many **built-in methods** that allow you to manipulate and query strings. Here's a curated list of commonly used ones:

#### **A. Case Conversion**

**There are generally three methods for case conversion in python: `upper()`, `lower()`, `.title()`**

````python
Lower_example = "Data Science"
print(Lower_example.lower())
````

````python
Upper_example = "data science"
print(Upper_example.upper())
````

````python
Title_case_example = "data science"
print(Title_case_example.title())
````

#### **B. Search and Replace**

You can search using `find()` method and you can replace using the `replace()` method.


````python
# To demonstrate find()

Story = "I am a Data Science mentor at Colab Innivation Hub"
Story.find("Science")
````

````python
# To demonstrate replace()

Story = "I am a Data Science mentor at Colab Innivation Hub"
Story.replace("Science", "Ananlyst")
````

### **2. String Formatting**


**String formatting is the process of dynamically inserting variables or values into a string template to produce a complete, human-readable sentence or output. Python offers several ways to embed variables or values inside strings:**


#### **A. F-strings**


````python
Bio_Narrator = "My name is James and I am a Data Science mentor working at Colab"
print(Bio_Narrator)
````

````python
# Using f string for dynamic data passing

Occupation = "Research Scientist"
Company = "OpenAI"

Improved_Bio_Narrator = f"My name is James and I an a {Occupation} working at {Company}"
print(Improved_Bio_Narrator)
````

#### **B. `str.format()` method**


````python
# Using .format() for dynamic data passing

Occupation = "Machine Learning Scientist"
Company = "DeepSeek"

Improved_Bio_Narrator = "My name is James and I an a {} working at {}".format(Occupation, Company)
print(Improved_Bio_Narrator)
````

###  **3. String Concatenation**

**String concatenation is the process of joining two or more strings together to form a single string. This can be done using the `+` method**


````python
# Demonstrating concatenation

Name = "Tinubu"
Surname = "Bola"

Full_name = Surname  +  Name
print(Full_name)
````

````python
Full_name = Surname + " " + Name
print(Full_name)
````

### **4. String Slicing / indexing**

Slicing allows you to extract parts of a string using `[start:stop:step]` notation.

#### **A. Basic Slicing**


````python
# Demonstarting slicing

Name = "Tinubu"
Surname = "Bola"

Full_name = Surname  + " " + Name
print(Full_name)
````

````python
# To pick the name Bola

X = Full_name[0:5]
X
````

````python
# To pick the name Bola

Y = Full_name[0:4]
Y
````

````python
# To pick the name Bola

Y = Full_name[0:3]
Y
````

````python
# To pick Tinubu

Z = Full_name[5:11]
Z
````

````python
# another method for the above
P = Full_name[5:]
P
````

#### **B. Step Slicing**

````python
Full_name
````

````python
Full_name[0:11:2]
````

````python
# let's slice through and pick numbers with 2 step

Numbers_of_interest = 0123456789
Numbers_of_interest[0:10:2]
````
