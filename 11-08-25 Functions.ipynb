#### **Problem Statement**

You are working as a **data analyst** for a global **health and fitness startup**. Your team is studying **average daily calorie intake** across different countries to design personalized nutrition plans.

The dataset is stored as a **dictionary**, where each **country** is a key and its **average daily calorie intake per person** is the value:

```python
avg_calories = {
    "USA": 3700.5,
    "Japan": 2720.3,
    "Nigeria": 3015.7,
    "Brazil": 3201.4,
    "Germany": 3770.5
}
```

Your tasks are:

### **1. Basic Operations**

* Use `min()` and `max()` to find the **country** with the **lowest** and **highest** calorie intake.
* Use `abs()` to find the **absolute difference** in calorie intake between the country with the highest and lowest average.
* Use `pow()` to calculate the **square** of the highest calorie intake (for a hypothetical “calorie impact score”).
* Use `round()` to **round** the global average calorie intake to **two decimal places**.

### **2. Aggregate Functions**

* Use `sum()` to compute the **total** calorie intake across all countries, then divide by the number of countries to find the **global average** (without using `statistics.mean`).

### **3. Special Math Functions** *(Import from `math`)*

* Use `ceil()` to round **up** the global average calorie intake.
* Use `floor()` to round **down** the global average calorie intake.
* Use `sqrt()` to find the **square root** of the lowest calorie intake (a hypothetical “nutrition baseline factor”).

````python
avg_calories = {
    "USA": 3700.5,
    "Japan": 2720.3,
    "Nigeria": 3015.7,
    "Brazil": 4201.4,
    "Germany": 3770.5
}
````

````python
import math
ceil(avg_calories)
````

````python
# Let's get the highest intake

max(avg_calories)
````

````python
avg_calories.get("USA")            
````

````python
# Let's get the highest intake value using the values instead of key

max(avg_calories, key=avg_calories.get)
````

````python
# Let's get the minimum intake value using the values instead of key

min(avg_calories, key=avg_calories.get)
````

````python
# Name the country with maximum intake

Country_max_cal_intake = max(avg_calories, key=avg_calories.get)
Country_max_cal_intake
````

````python
# Name the country with minimum intake

Country_min_cal_intake = min(avg_calories, key=avg_calories.get)
Country_min_cal_intake
````

````python
# Get the values by using Dictionary_name[key]

max_country_cal_score = avg_calories[Country_max_cal_intake]
max_country_cal_score
````

````python
# Get the values by using Dictionary_name[key]

min_country_cal_score=avg_calories[Country_min_cal_intake]
min_country_cal_score
````

````python
difference = min_country_cal_score - max_country_cal_score
abs_difference = abs(difference)
print(f"The absolute caloris intake between the max and min country is {abs_difference}")
````
