#### List Comprehension

```python
# Use list comprehension to remove two elements at the same time
```

```python
bm = ['BMW']
```

```python
type(bm)
```

```python
daddy_girl = [x for x in daddy_girl if x not in bm]
```

```python
daddy_girl
```

```python
num_list = [3,4,5,6] # Another example of list comprehension
```

```python
[x ** 2 for x in num_list]
```

```python
for x in num_list:
    print(x ** 2)
```
