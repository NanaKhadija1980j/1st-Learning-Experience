## **Control Structures in Python Programming --> Loops**

A **loop** is a programming construct that allows you to repeatedly execute a block of code as long as a specified condition is met. Loops are used to automate repetitive tasks, iterate over sequences (like lists, strings, or ranges), and process data efficiently.


### **Types of Loops in Python**

Python provides two main types of loops:

1. **`for` Loop**: The for loop is used to iterate over a sequence (e.g., a list, tuple, string, or range) and execute a block of code for each item in the sequence.

2. **`while` Loop**: The `while` loop repeatedly executes a block of code as long as a specified condition is `True`. It is useful when the number of iterations is not known in advance.


**Today we will be looking at `for loops` and common ways they are being used in Python programming.**


### **Basic Syntax of a For Loop**  

```python
for variable in sequence:
    # Code block to execute
```

- `variable` → A temporary variable that takes the value of each item in `sequence`.  
- `sequence` → A collection (e.g., list, tuple, string, range).  
- **Indentation** → Code inside the loop must be indented.  

### **1. Looping Through Strings**
A string is a sequence of characters. You can use a `for` loop to iterate over each character in the string.

````python
Class = "Data Science"

for alphabet in Class:
    print(alphabet)
````

### **2. Looping Through `range()`**
The `range()` function generates and loop through a sequence of numbers. which is commonly used in `for` loops with:

- 1. With basic `range`
- 2. With `start` and `stop`
- 3. With `step`


### Syntax:

```python
for i in range(start, stop, step):
    # Code block
```

````python
# Count 0 to 10

for number in range(11):
    print(number)
````

````python
# Count from 1 to 20

for number in range(20):
    print(number+1)
````

````python
# Count 1 to 20

for number in range(1,21):
    print(number)
````

##### **Using `range(start, stop, step)`**

````python
# Count from 1 to 50 with only even numbers

for number in range(0,50,2):
    print(number)
````

````python
for number in range(50):
    if number % 2 == 0:
        print(number)
````

A list is a collection of items. You can use a `for` loop to iterate over each item in the list.

### **4. Looping Through Tuples**
A tuple is similar to a list but immutable. You can loop through tuples in the same way as lists.

````python
# Let's  biold a password length checker 

Password_length_Checker = ("Abcde12345", "bcdef23456", "cdefg34567", "def456", "efg5678")

for i in Password_length_Checker:
    if len(i) != 10:
        print(f"Sorry the password {i} is not accepted, we need a password of 10 characters")
    else:
        print(f"The password {i} meets the reauirements")
````

1. **Looping through keys**

````python
# Accesing key using for loops

for key in Data_Science_Team:
    print(key)
````

## **Advanced `for` Loop Techniques**

### **6. Using `enumerate()` in Loops**
The `enumerate()` function adds a counter to an iterable, allowing you to loop through both the index and the value of each item.

#### Syntax:

```python
for index, value in enumerate(iterable):
    # Your code here
```

* `iterable`: the sequence you're looping through (e.g., list, string).

````python
# Picking dictionary like behavior from list

List_DS_Students = ["Paul", "Abdul", "Nasir", "Odinaka"]
List_DS_Students[0]
````

````python
# Picking dictionary like behavior from tuple

Tuple_DS_Students = ("Paul", "Abdul", "Nasir", "Odinaka")
Tuple_DS_Students[1]
````

````python
# Using emunrate() function

for index, value in enumerate(List_DS_Students):
    print(index, value)
````

````python
for index, value in enumerate(Tuple_DS_Students):
    print(index, value)
````

````python
List_of_Salary = [20000, 15000, 18000, 25000, 30000]
List_of_Salary[0] = 200
List_of_Salary[0]
````

````python
List_of_Salary = [20000, 15000, 18000, 25000, 30000]

for index, value in enumerate(List_of_Salary):
    if index >1:
        List_of_Salary[index] = List_of_Salary[index] + 1000
        
````

#### **Custom enumerate() with Start Index:**

#### Syntax:

```python
for index, value in enumerate(iterable, start=0):
    # Your code here
```

* `iterable`: the sequence you're looping through (e.g., list, string).
* `start`: optional; the starting index (default is `0`).

### **7. Using `zip()` for Parallel Iteration**
The `zip()` function allows you to iterate over multiple sequences (e.g., lists, tuples) in parallel. It pairs corresponding elements from each sequence.

#### **Syntax**
```python
for item1, item2, ... in zip(sequence1, sequence2, ...):
    # Code to execute for each pair of items
```

**Note:** If the sequences are of unequal length, `zip()` stops at the shortest sequence.

````python
List_DS_Students = ["Paul", "Abdul", "Nasir", "Odinaka", "Cynthia"]

List_of_Salary = [20000, 15000, 18000, 25000, 30000]
````

````python
for name, salary in zip(List_DS_Students, List_of_Salary):
    print(f"{name} receive a salary of {salary} naira")
````

### **8. Intro to List and Dictionary Comprehension in Python**

**Comprehensions** are a concise and readable way to **create new lists, dictionaries, or sets** from existing iterables (like lists, strings, or ranges). They replace loops with a single line of expressive syntax.

Python supports:

* **List comprehension**: for creating lists
* **Dictionary comprehension**: for creating dictionaries

They make your code **shorter**, **cleaner**, and often **faster**.

#### **List Comprehension**

##### Syntax:

```python
[expression for item in iterable if condition]
```

#### **Dictionary Comprehension**

##### Syntax:

```python
{key_expression: value_expression for item in iterable if condition}
```
