# **Data Structures in Python  --> Set**

- A **set** is an unordered collection of unique elements.
- Sets are mutable, meaning you can add or remove elements.
- Sets do not allow duplicate values.

**We will be looking how to do the following:**
- Creating a set
- Accessing set members
- Changing items
- Adding items
- Removing items
- Deleting a set
- Set Operation: `Union, Intersect, Difference, Subset, and Super set`

```python
DS_Students_Names = {"Abdullahi", "Amanda", "Ifeoma", "Paul",  "Chioma", "Paul"}
```

```python
# Checking for memebers

"James" in DS_Students_Names
```

```python
"Ifeoma" in DS_Students_Names
```

### **3. Modifying a Set**

**Changing items**

Sets are mutable, but since they are unordered, you cannot change a specific element directly. Instead, you can `add` or `delete` elements.

#### **1. Adding Items**
You can add a single element using the `add()` method or multiple elements using the `update([])` method.


```python
# Adding a single item

Nigeria_as_Country_of_residence.add("Afolabi")
```

```python
Nigeria_as_Country_of_residence
```

```python
# Adding multip[le items

Nigeria_as_Country_of_residence.update(["Jedida", "Patience"])
```

```python
Nigeria_as_Country_of_residence
```

```python
# Let's correct the spelling of jedida to jedidah

Nigeria_as_Country_of_residence.update(["Jedidah", "Patience"])
```

```python
Nigeria_as_Country_of_residence
```


#### **2. Removing Items**
You can remove elements using the remove(), and discard() methods.

```python
# Let's remove the wrongly spelt name = jedida

Nigeria_as_Country_of_residence.discard("Jedida")
Nigeria_as_Country_of_residence
```

### Demonstrate .remove() versus .discard()

```python
# Demonstrating the difference between .remove() and discard

Nigeria_as_Country_of_residence.remove("Jedida")
Nigeria_as_Country_of_residence
```

```python
# With discard there is no error

Nigeria_as_Country_of_residence.discard("Jedida")
```

```python
# With remove we get an error

Nigeria_as_Country_of_residence.remove("Jedida")
```

### Notice that `.remove()`  will raise an error if the values is not in the set but `.discard()` will not raise an error.

### Removing more than one item using `difference_update({})`

```python
Nigeria_as_Country_of_residence.difference_update({"Amanda", "Chioma"})
```

```python
Nigeria_as_Country_of_residence
```

**Sometimes you will want to empty a set or even completely remove it. To empty a set, you can use `.clear()`**

### **Set Operations**

We will be looking at three major set operations: Union, Intersection, Difference.

#### **1. Union (`union()`)**
The `union()` method combines two sets and returns a new set containing all unique elements from both sets.

- Syntaxt 1: union_set = set1.union(set2)
- Syntaxt 2: union_set = set1 | set2

```python
Wala_skill_set = {"Football", "Data Scientist", "Graphics", "Python"}
Ifeoma_skill_set = {"Cook", "Data Scientist", "Fashio Design", "Python"}

Union_Set = Wala_skill_set.union(Ifeoma_skill_set)
Union_Set
```

```python
Union_Set_B = Wala_skill_set | Ifeoma_skill_set
Union_Set_B
```

#### **2. Intersection (`intersection()`)**
The `intersection()` method returns a new set with elements that are common to both sets.

- Syntax 1: intersection_set = set1.intersection(set2)
- Syntax 2: intersection_set = set1 & set2

```python
Wala_skill_set
```

```python
Ifeoma_skill_set
```

```python
Intersection_Set = Wala_skill_set.intersection(Ifeoma_skill_set)
Intersection_Set
```

```python
Intersection_Set_B = Wala_skill_set & Ifeoma_skill_set
Intersection_Set_B
```

### **3. Difference (`difference()`)**
The `difference()` method returns a new set with elements that are in the first set but not in the second set.

- Syntax 1: difference_set = set1.difference(set2)
- Syntax 2: difference_set = set1 - set2

```python
# What set Wala_skill_set has that set Ifeoma_skill_set does not have

Intersection_Set_Diff = Wala_skill_set.difference(Ifeoma_skill_set)
Intersection_Set_Diff
```

```python
# What set Wala_skill_set has that set Ifeoma_skill_set does not have

Intersection_Set_Diff_B = Ifeoma_skill_set.difference(Wala_skill_set)
Intersection_Set_Diff_B
```

```python
# What set Wala_skill_set has that set Ifeoma_skill_set does not have

Intersection_Set_Diff_C = Wala_skill_set - Ifeoma_skill_set
Intersection_Set_Diff_C
```

```python
# What set Wala_skill_set has that set Ifeoma_skill_set does not have

Intersection_Set_Diff_D = Ifeoma_skill_set - Wala_skill_set
Intersection_Set_Diff_D
```

```python
# Using the update method with tuple

A = {"abc", "def", "ghi"}
A.update(("Yes", "No"))
A
```

```python
# Using the update method with list

B = {123, 456, 789}
B.update([101, 102])
B
```

### Demostarting Sub-set and Super-set

```python
Wale_skill_set = {"Python", "Teacher", "Cook", "Fashio Designer"}
Amanda_skill_set = {"Python", "Cook", "Fashio Designer"}
```

```python
# Let's check if Wale_skill_set is a sub set of Amanda_skill_set

Wale_skill_set.issubset(Amanda_skill_set)
```

```python
# Let's check if Amanda is a superset of Wale_skill_set

Amanda_skill_set.issuperset(Wale_skill_set)
```

```python
# Let's check if Wale_skill_set is a superset of Amanda_skill_set

Wale_skill_set.issuperset(Amanda_skill_set)
```
